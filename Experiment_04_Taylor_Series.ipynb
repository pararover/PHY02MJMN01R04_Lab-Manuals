# Experiment 4

## Convergence of Taylor series

### Objectives

1. To compute the first $N=5,6,8,10$ terms of the Taylor series for $\cos{x}$ from $x=0$ to $x=8$, given as
\begin{align*}
	T(x)=\sum_{n=0}^{\infty}{\dfrac{(-1)^n}{(2n)!}x^{2n}}
\end{align*}
1. To plot and compare the convergence of the Taylor series partial sums with the analytical result of the $\cos$ function.

### Importing required libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from math import factorial
```

### Defining analytical cosine function

```python
def cos_x(x):
    return np.cos(x)
```

### Defining Taylor series of partial sum

```python
def taylor_series_cos(x, N):
    partial_sum = 0
    for n in range(0, N + 1):
        partial_sum += ((-1)**n / factorial(2 * n)) * (x**(2 * n))
    return partial_sum
```

### Generating $x$ values

```python
xi, xf = 0, 8
N = 500
x_values = np.linspace(xi, xf, N)
```

### Computing analytical values of cosine

```python
y_cos = cos_x(x_values)
```

### Setting number of terms in Taylor series

```python
Ns = [5, 6, 8, 10]
```

### Plotting convergence of Taylor series

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
axes = axes.flatten() # Flatten the 2x2 array of axes for easier iteration

for i, N in enumerate(Ns):
    y_taylor = taylor_series_cos(x_values, N)
    axes[i].plot(x_values, y_cos, label='cos(x)', color='blue', linestyle='-')
    axes[i].plot(x_values, y_taylor, label=f'Taylor Series (N={N})', color='red', linestyle='--')
    axes[i].set_title(f'Comparison: cos(x) vs. Taylor Series (N={N})')
    axes[i].set_xlabel('x')
    axes[i].set_ylabel('y')
    axes[i].legend()
    axes[i].grid(True)
    axes[i].set_ylim(-2, 2)

plt.tight_layout()
plt.show()
```
