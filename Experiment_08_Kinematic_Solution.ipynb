# Experiment 8

## Kinematic solution of time-dependent force

### Objectives

Given a particle of mass $m=1$ kg under the influence of the following time-dependent force
\begin{align*}
  F=\alpha+\beta t
\end{align*}
1. To numerically compute the velocity $v(t)$ and position $x(t)$ if the particle starts from position $x_0=1$ m with velocity $v_0=-5$ m/s, taking $\alpha=0.5$ N and $\beta=0.1$ N/s.
1. To plot the velocity $v$ and position $x$ for $0\leqslant t\leqslant 20$ s.

### Importing required libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
```

### Defining the ODE for the system

```python
def odes_scenario1(t, y, m, alpha, beta):
    # y[0] is x (position)
    # y[1] is v (velocity)
    x, v = y

    F = alpha + beta * t
    dvdt = F / m
    dxdt = v
    return [dxdt, dvdt]
```

### Setting constant values

```python
m = 1.0       # mass
alpha = 0.5   # constant part of force
beta = 0.1    # coefficient for time-dependent part of force
```

### Setting initial conditions

```python
x0 = 1.0
v0 = -5.0
y0 = [x0, v0]
```

### Creating time instants for computation

```python
t_span = (0, 20)
t_eval = np.linspace(t_span[0], t_span[1], 500)
```

### Solving the ODE of the system

```python
solution = solve_ivp(
    odes_scenario1,
    t_span,
    y0,
    args=(m, alpha, beta),
    t_eval=t_eval,
    rtol=1e-6, atol=1e-9 # Relative and absolute tolerances
)
```

### Extraction of results

```python
t1 = solution.t
x1 = solution.y[0]
v1 = solution.y[1]
```

### Plotting velocity and position graphs

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(t1, v1, label='Velocity v(t)')
plt.xlabel('Time (t)')
plt.ylabel('Velocity')
plt.title(r'Scenario 1: Velocity vs. Time ($F = \alpha + \beta t$)')
plt.grid(True)
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(t1, x1, label='Position x(t)', color='red')
plt.xlabel('Time (t)')
plt.ylabel('Position')
plt.title(r'Scenario 1: Position vs. Time ($F = \alpha + \beta t$)')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```
