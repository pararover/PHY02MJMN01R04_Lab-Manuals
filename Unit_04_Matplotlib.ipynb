# Visualization using the `Matplotlib` library

* `Matplotlib` is a comprehensive, open-source Python library for creating static, animated, and interactive visualizations.
* `pyplot` is a specific module within the `Matplotlib` library that provides a simple, state-based interface for generating many different types of plots and graphical representations of data, using just a few lines of code.
* The user has full control of line styles, font properties, axes properties, etc.
* Usually data points for the plotting functions are supplied as Python lists or as `NumPy` arrays.

## Importing the `matplotlib.pyplot` module

````python
import matplotlib.pyplot as plt
````

## Creating 2D plots

General syntax of the `plot()` function:
```markdown
plt.plot(x, y, [fmt], **kwargs)
```
Here `[fmt]` is an optional string format and `**kwargs` are optional keyword arguments specifying the properties of the plot.

* To improve the appearance of the plot, we use `pyplot` functions to add a figure title, legend, gridlines, etc.
* The `plt.show()` function displays the figure.

### Example: Position-time graph of ball thrown upwards

````python
import numpy as np
import matplotlib.pyplot as plt
````

````python
vo, g = 5, 9.8 # define initial speed and acceleration due to gravity
````

````python
t = np.linspace(0,1.03,100)       # Create a sequence of time instants
y = [vo*T - g*T**2/2 for T in t]  # Create a sequence of positions
````

````python
plt.plot(t, y)
````

````python
plt.plot(t, y);
````

````python
plt.plot(t, y)
plt.show()
````

````python
plt.plot(t, y)

plt.title('Ball thrown straight upwards') # Add title to plot
plt.ylabel('Vertical distance y(t)')      # Add labels for x and y axes
plt.xlabel('Time t[s]')

ymax = max(y)                              # Find position ymax
tmax = t[np.argmax(y)]                     # Find time tmax
print('Max height attained =',f'{ymax:.4f}', ' m')
print('Time to reach ymax =',f'{tmax:.4f}',' s')

plt.show()
````

## Creating 3D plots

### Example: Plot of the surface $z=3+x+y$

````python
import numpy as np
import matplotlib.pyplot as plt
````

````python
# Surface plot for z = 3 + x + y
x = np.arange(-3, 3, 0.6) # Grid of points on x-axis
X, Y = np.meshgrid(x, x)  # Grid of points on xy-plane
Z = 3 + X + Y             # Values of z for points on the xy-grid
````

````python
# Plot 3D surface
# Define the objects fig and axes for 3D plotting
fig = plt.figure()
axes = fig.add_subplot(projection='3d')

axes.plot_surface(X, Y, Z, color='skyblue')

axes.set_xlabel('X')  # Labels for x, y, z axes
axes.set_ylabel('Y')
axes.set_zlabel('Z')
axes.text(-2.6, 2, 6, 'z = f(x,y) = 3 + x + y') # Annotate the plot
plt.show()
````

````python
# Plot 3D wireframe
# Define the objects fig and axes for 3D plotting
fig = plt.figure()
axes = fig.add_subplot(projection='3d')

axes.plot_wireframe(X, Y, Z, color='green')

axes.set_xlabel('X')  # Labels for x, y, z axes
axes.set_ylabel('Y')
axes.set_zlabel('Z')
axes.text(-2.6, 2, 6, 'z = f(x,y) = 3 + x + y') # Annotate the plot
plt.show()
````
