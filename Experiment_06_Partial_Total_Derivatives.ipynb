# Experiment 6

## Symbolic partial and total differentiation

### Objectives

Given the following Gaussian-weighted quadratic function
$$f=ax^2e^{-\lambda(x^2+y^2)}$$
where $a$ and $\lambda$ are constants and $x=x(t)$ and $y=y(t)$. Use symbolic computations in Python
1. To compute $\left(\dfrac{\partial f}{\partial x}\right)_y$ and $\left(\dfrac{\partial f}{\partial y}\right)_x$.
1. To show that $\dfrac{\partial f}{\partial t}$ explicitly.
1. To compute the total derivative $\dfrac{\mathrm{d}f}{\mathrm{d}t}$ using the implicit dependence of $f$ on $t$ through $x(t)$ and $y(t)$.

### Importing required libraries

```python
from sympy import symbols, diff, exp, Function
```

### Defining symbolic variables

```python
a, lambda_ = symbols('a lambda') # Using lambda_ to avoid conflict with Python keyword
x, y, t = symbols('x y t')
```

### Defining Gaussian-weighted function

```python
f = a * x**2 * exp(-lambda_ * (x**2 + y**2))

print("Given function f:")
display(f)
```

### Computing $\left(\dfrac{\partial f}{\partial x}\right)_y$

```python
df_dx = diff(f, x)
display(df_dx)
```

### Computing $\left(\dfrac{\partial f}{\partial y}\right)_x$

```python
df_dy = diff(f, y)
display(df_dy)
```

### Computing $\dfrac{\partial f}{\partial t}$

```python
df_dt_explicit = diff(f, t)
display(df_dt_explicit)
```

### Computing $\dfrac{\mathrm{d}f}{\mathrm{d}t}$
The total derivative is computed, taking into account the implicit dependence of $x$ and $y$ on $t$, as follows:
$$\dfrac{\mathrm{d}f}{\mathrm{d}t}=\dfrac{\partial f}{\partial x}\dfrac{\mathrm{d}x}{\mathrm{d}t}+\dfrac{\partial f}{\partial y}\dfrac{\mathrm{d}y}{\mathrm{d}t}$$

```python
x_func = Function('x')(t)
y_func = Function('y')(t)

f_t = f.subs({x: x_func, y: y_func})

df_dt_total = diff(f_t, t)
display(df_dt_total)
```
