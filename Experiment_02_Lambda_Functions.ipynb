# Experiment 2

## Applications of lambda functions

### Objectives

1. To plot the functions $f(x)=\dfrac{x^2}{10}$ and $g(x)=\sin{x}$ using lambda
functions over the range $0\leqslant x\leqslant 6$.
1. To create a grid of three subplots, one for each function and a third showing both functions.

### Importing required libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Defining the lambda functions

```python
f = lambda x: x**2 / 10
g = lambda x: np.sin(x)
```

### Generating $x$ values

```python
N = 100
x = np.linspace(0, 6, N)
```

### Plotting the lambda functions

#### Quadratic function

```python
plt.figure(figsize=(6, 5))
plt.plot(x, f(x), color='blue', linestyle='-.')
plt.title(r'Plot of $f(x)=\dfrac{x^2}{10}$')
plt.xlabel(r'$x$', fontsize=12)
plt.ylabel(r'$f(x)$', fontsize=12)
plt.grid(True)
plt.show()
```

#### Trigonometric function

```python
plt.figure(figsize=(6, 5))
plt.plot(x, g(x), color='red', linestyle='--')
plt.title(r'Plot of $g(x)=\sin{x}$')
plt.xlabel(r'$x$', fontsize=12)
plt.ylabel(r'$g(x)$', fontsize=12)
plt.grid(True)
plt.show()
```

### Creating a grid of sub-plots

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Plotting f(x) = x^2/10
axes[0].plot(x, f(x), color='blue', linestyle='-.', label=r'$f(x)=\dfrac{x^2}{10}$')
axes[0].set_title(r'Plot of $f(x)=\dfrac{x^2}{10}$')
axes[0].set_xlabel(r'$x$', fontsize=12)
axes[0].set_ylabel(r'$f(x)$', fontsize=12)
axes[0].grid(True)
axes[0].legend()

# Plotting g(x) = sin x
axes[1].plot(x, g(x), color='red', linestyle='--', label=r'$f(x)=\sin{x}$')
axes[1].set_title(r'Plot of $g(x)=\sin{x}$')
axes[1].set_xlabel(r'$x$', fontsize=12)
axes[1].set_ylabel(r'$g(x)$', fontsize=12)
axes[1].grid(True)
axes[1].legend()

# Plotting both functions together
axes[2].plot(x, f(x), color='blue', linestyle='-.', label=r'$f(x)=\dfrac{x^2}{10}$')
axes[2].plot(x, g(x), color='red', linestyle='--', label=r'$f(x)=\sin{x}$')
axes[2].set_title('Plot of $f(x)$ and $g(x)$')
axes[2].set_xlabel(r'$x$', fontsize=12)
axes[2].set_ylabel(r'$f(x)$, $g(x)$', fontsize=12)
axes[2].grid(True)
axes[2].legend()
```
