## Loop statement: `for`

Python `for` loops are used for iterating over sequences like lists, tuples, strings and ranges.

* A for loop allows you to apply the same operation to every item within the loop.
* Using a for loop avoids the need to manually manage the index.
* A for loop can iterate over any iterable object, such as a dictionary, list or custom iterator.

In Python, an
iterable is any object capable of returning its members one at a time.
- allows to loop over it using a `for` loop, list comprehension, or other iteration tools
- Common iterables: `list`, `tuple`, `str`, `dict`, `set`, `range` objects

General syntax for `for` statement:
```markdown
for var in iterable:
  # statements
```

````python
s = "Scientific Analysis"
for i in s:
    print(i)
````

````python
for x in [1, 2, 3, 4, 5]:
  e = 2 * x
  o = 2 * x + 1
  print(f"{e} is even.")
  print(f"{o} is odd.")
````

````python
for i in range(5, 15, 2):
  print(i)
````

## Combining conditionals with loops

````python
original_string = 'computational skills in scientific analysis'
modified_string = ''
vowels = ['a', 'e', 'i', 'o', 'u']
for i in original_string:
  if i in vowels:
    continue
  modified_string = modified_string + i

print(original_string)
print(modified_string)
````

````python
tpcssap = ["training", "programme", "on", "computational", "skills", "in", "scientific", "analysis", "using", "python"]

for index, word in enumerate(tpcssap):
    print(index, word)
````

````python
for i in range(1, 4):
  for j in range(1, 4):
    print(i, j)
````

# Functions

In Python functions refer to a block of statements that perform a specific task.

The idea is to put some commonly or repeatedly done task together and make a function so that instead of writing the same code again and again for different inputs, we can do the function calls to reuse code contained in it over and over again.

In Python, functions are defined using the `def` keyword.

General syntax of Python functions:
```markdown
def function_name(parameters):
  """Docstring""" [optional]
  # body of the function
  return expression [optional]
```

````python
def evenOdd(x):
  if (x % 2 == 0):
    return "Even"
  else:
    return "Odd"
````

````python
evenOdd(17)
````

````python
evenOdd(348)
````

## Default arguments
A default argument is a parameter that assumes a default value if a value is not provided in the function call for that argument.

````python
def demo_function(x, y=50):
    print("x: ", x)
    print("y: ", y)
````

````python
demo_function(10)
````

````python
demo_function(10, 19)
````

## Positional arguments
In positional arguments, values are assigned to parameters based on their order in the function call.

````python
def introduction(name, age):
  print("My name is", name)
  print("My age is", age)
````

````python
introduction("James", 35)
````

````python
introduction(35, "James")
````

## Keyword arguments
In keyword arguments, values are passed by explicitly specifying the parameter names, so the order doesn’t matter.

````python
def full_name(firstname, lastname):
  print(firstname, lastname)
````

````python
full_name("Physics", "Department")
````

````python
full_name("Department", "Physics")
````

````python
full_name(firstname="Physics", lastname="Department")
````

````python
full_name(lastname="Department", firstname="Physics")
````

## `return` statement
In Python, the `return` statement ends a function and sends a value back to the caller. It can return any data type, multiple values (packed into a tuple), or None if no value is given.

````python
def square_root(number):
  sq_root = number**0.5
  return sq_root
````

````python
x = 169
y = square_root(x)

print("x =", x)
print("Square root of x=", y)
````

## Lambda functions
A lambda function is a small anonymous function.

A lambda function can take any number of arguments, but can only have one expression.

General syntax for lambda functions:
```markdown
lambda [arguments] : [expression]
```

````python
x = lambda a : a + 10   # A lambda function with one argument

print(x(5))
print(x(-15))
````

````python
y = lambda a, b : a + b**2

print(y(5, 6))
print(y(1, -4))
````

### Use of lambda functions

````python
def myfunc(n):
  return lambda a : a * n
````

````python
mydoubler = myfunc(2)

print(mydoubler(11))
````

````python
mytripler = myfunc(3)

print(mytripler(11))
````

# Importing libraries & packages

Python packages are a way to organize and structure code by grouping related modules into directories.
* A package is essentially a folder that contains an __init__.py file and one or more Python files (modules).
* Allows modules to be easily shared and distributed across different applications.

>>**Module:** A single Python file (.py) containing code.

>>**Package:** A collection of modules (folders with an `__init__.py` file).

>>**Library:** A large collection of packages and modules for a specific purpose (e.g., `matplotlib`, `numpy`, `sympy`, `scipy`)

## Importing entire modules
