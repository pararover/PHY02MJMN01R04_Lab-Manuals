# Numerical computations using the `NumPy` library

* `NumPy` is the core Python library for numerical computing.
* `NumPy` supports operations on compound data types like arrays and matrices.
* The first thing to learn is how to create arrays and matrices using the `NumPy` library.

## Importing the `NumPy` library

````python
import numpy as np
````

## Creating `NumPy` arrays

### Creating N-dimensional `NumPy` array from sequence `a`
```markdown
np.array(a)
```

````python
a = np.array([1, 5, 2, 7, 4])

print(a)
````

### Creating an evenly spaced 1D array from `start` to `stop`, excluding the `stop` value
```markdown
np.arange(start, stop, step)
```

````python
b = np.arange(5, 10, 0.5)

print(b)
````

### Creating a 1D array with length equal to `N`, including the `start` and `stop` values
```markdown
np.linspace(start, stop, N)
```

````python
c = np.linspace(-1, 1, 11)

print(c)
````

### Creating array of given shape and type, filled with 0's
```markdown
np.zeros(shape)
```

````python
d = np.zeros(5)

print(d)
````

````python
e = np.zeros((3, 4))

print(e)
````

### Creating array of given shape and type, filled with 1's
```markdown
np.ones(shape)
```

````python
f = np.ones(5)

print(f)
````

````python
g = np.ones((3, 4))

print(g)
````
