# Symbolic Computations using the `SymPy` Library

* Symbolic computation deals with the computation of mathematical objects symbolically.
  - This means that the mathematical objects are represented exactly, not approximately, and mathematical expressions with unevaluated variables are left in symbolic form.
* `SymPy` (short for `Sym`bolic `Py`thon) is a core Python library for symbolic mathematics.
* It enables exact computation with algebraic expressions, calculus, matrices.
* `SymPy` introduces symbolic objects like variables and expressions, allowing manipulation and transformation using algebraic rules.

## Symbolic integration

````python
integrand = sym.exp(x)*sym.sin(x)
integrand
````

### Indefinite integration

\begin{align*}
  \int{e^x\sin{x}\,\mathrm{d}x}
\end{align*}

````python
sym.integrate(integrand, x)
````

### Definite integration

\begin{align*}
  \int\limits_{0}^{1}{e^x\sin{x}\,\mathrm{d}x}
\end{align*}

````python
sym.integrate(integrand, (x, 0, 1))
````

````python
sym.integrate(integrand, (x, 0, 1)).simplify()
````

````python
sym.integrate(integrand, (x, 0, 1)).evalf()
````

## Solving differential equations

Example: To solve the differential equation
$$\dfrac{\mathrm{d}^2y}{\mathrm{d}t^2}-y=e^t$$

````python
t = sym.symbols('t')
y = sym.Function('y')
````

````python
y(t)
````

````python
y(t).diff(t)
````

````python
y(t).diff(t, t)
````

````python
LHS = y(t).diff(t, t) - y(t)
LHS
````

````python
RHS = sym.exp(t)
RHS
````

````python
ODE = sym.Eq(LHS, RHS)
ODE
````

````python
sym.dsolve( ODE, y(t) )
````

## The `lambdify()` function

* The `subs()` and `evalf()` functions evaluate an expression at a specific point.
* To evaluate an expression at any arbitrary point, we make use of the `lambdify()` function to convert a `SymPy` expression into a `lambda` function.

General syntax for the `lambda()` function:
```markdown
sympy.lambdify(arguments, expression)
```

Example: For a hydrogen-like ion with atomic number $Z$, the wavelength $\lambda$ of the photon emitted when the lone electron makes a transition from a higher energy level $n_2$ to a lower energy level $n_1$ is given by Rydberg's formula:
$$\dfrac{1}{\lambda}=RZ^2\left[ \dfrac{1}{n_1^2}-\dfrac{1}{n_2^2} \right]$$
where the Rydberg's constant is $R=1.097\times 10^{-2}$ nm$^{-2}$.

Ref: https://en.wikipedia.org/wiki/Rydberg_formula

````python
R, Z, n1, n2 = sym.symbols('R Z n_1 n_2')

RH = 1.097E-2 # in nanometres
````

````python
type(R)
````

````python
type(RH)
````

````python
lambda_inverse = R * Z**2 * ( (1/n1**2) - (1/n2**2) )
lambda_inverse
````

````python
rydberg_lambda = sym.lambdify( [n1, n2, Z], (1/lambda_inverse).subs([(R, RH)]) )
````

````python
type(rydberg_lambda)
````

````python
rydberg_lambda(1, 2, 1)
````

````python
rydberg_lambda(1, 3, 1)
````

````python
(1/lambda_inverse).subs([(n1, 1), (Z, 1), (R, RH)])
````

````python
sym.limit( (1/lambda_inverse).subs([(n1, 1), (Z, 1), (R, RH)]), n2, sym.oo )
````

````python
# Lyman convergence
sym.limit( (1/lambda_inverse).subs([(n1, 1), (Z, 1), (R, RH)]), n2, sym.oo ).evalf()
````

````python
# Balmer convergence
sym.limit( (1/lambda_inverse).subs([(n1, 2), (Z, 1), (R, RH)]), n2, sym.oo ).evalf()
````

````python
# Paschen convergence
sym.limit( (1/lambda_inverse).subs([(n1, 3), (Z, 1), (R, RH)]), n2, sym.oo ).evalf()
````

````python
# Brackett convergence
sym.limit( (1/lambda_inverse).subs([(n1, 4), (Z, 1), (R, RH)]), n2, sym.oo ).evalf()
````

````python
# Pfund convergence
sym.limit( (1/lambda_inverse).subs([(n1, 5), (Z, 1), (R, RH)]), n2, sym.oo ).evalf()
````

````python
# Humphreys convergence
sym.limit( (1/lambda_inverse).subs([(n1, 6), (Z, 1), (R, RH)]), n2, sym.oo ).evalf()
````
