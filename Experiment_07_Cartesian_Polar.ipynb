## Cartesian coordinates as functions of polar coordinates

### Objectives

The Cartesian coordinates are related to polar coordinates as
\begin{align*}
  x&=r\cos{\phi} \\
  y&=r\sin{\phi}
\end{align*}
1. To obtain the the expression of $y$ in terms of $x$ and $\phi$.
1. To symbolically calculate $\left( \dfrac{\partial y}{\partial\phi} \right)_x$.
1. To symbolically calculate $\left( \dfrac{\partial y}{\partial\phi} \right)_r$.

### Importing required libraries

```python
from sympy import symbols, cos, sin, diff, tan
```

### Defining symbolic variables

```python
r, phi, x_cart, y_cart = symbols('r phi x y')
```

### Setting up formulae relating Cartesian to polar coordinates

```python
x_polar = r * cos(phi)
y_polar = r * sin(phi)
```

```python
print("Given: x =")
display(x_polar)
print('\n')
print("Given: y =")
display(y_polar)
```

### Expressing $y$ in terms of $x$ and $\phi$

```python
# Express r in terms of x and phi
r_expr = x_cart / cos(phi)

# Substitute r into y to get y in terms of x and phi
y_as_f_of_x_phi = y_polar.subs(r, r_expr)

print("y expressed as a function of x and phi:")
display(y_as_f_of_x_phi)
```

### Computing $\left( \dfrac{\partial y}{\partial\phi} \right)_x$

```python
dy_dphi_x = diff(y_as_f_of_x_phi, phi)

print("\nThe partial derivative (dy/dphi)_x =")
display(dy_dphi_x)
```

### Computing $\left( \dfrac{\partial y}{\partial\phi} \right)_r$

```python
dy_dphi_r = diff(y_polar, phi)

print("The partial derivative (dy/dphi)_r =")
display(dy_dphi_r)
```
