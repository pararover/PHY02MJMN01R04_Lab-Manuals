## Symbolic Evaluation of Infinite Sum

### Objectives

To use the `Sum()` and `.doit()` methods of SymPy to evaluate the following infinite sums:
1. $\sum_{n=0}^{\infty}{\dfrac{(-1)^n}{(2n+1)!}x^{2n+1}}$
1. $\sum_{n=0}^{\infty}{\dfrac{1}{(2n)!}x^{2n}}$
1. $\sum_{n=0}^{\infty}{\dfrac{(-1)^n}{(2n)!}x^{2n}}$
1. $\sum_{k=0}^{\infty}{\dfrac{x^k}{k!}}$

### Importing necessary SymPy methods

```python
from sympy import symbols, Sum, oo, factorial, sin, cos, exp, log
from sympy.abc import n, k, x
```

### Taylor series expansion for $\sin{x}$
$$\sum_{n=0}^{\infty}{\dfrac{(-1)^n}{(2n+1)!}x^{2n+1}}$$

```python
sum_1 = Sum((-1)**n * x**(2*n + 1) / factorial(2*n + 1), (n, 0, oo))
result_1 = sum_1.doit()
print(f"The symbolic expression for the sum is: {sum_1}")
print(f"Symbolically valuated result: {result_1}")
```

### Taylor series expansion for $\cosh{x}$.
$$\sum_{n=0}^{\infty}{\dfrac{1}{(2n)!}x^{2n}}$$

```python
sum_2 = Sum(x**(2*n) / factorial(2*n), (n, 0, oo))
result_2 = sum_2.doit()
print(f"The symbolic expression for the sum is: {sum_2}")
print(f"Symbolically valuated result: {result_2}")
```

### Taylor series expansion for $\cos{x}$
$$\sum_{n=0}^{\infty}{\dfrac{(-1)^n}{(2n)!}x^{2n}}$$

```python
sum_3 = Sum((-1)**n * x**(2*n) / factorial(2*n), (n, 0, oo))
result_3 = sum_3.doit()
print(f"The symbolic expression for the sum is: {sum_3}")
print(f"Symbolically valuated result: {result_3}")
```

### Taylor series expansion for $e^x$
$$\sum_{k=0}^{\infty}{\dfrac{x^k}{k!}}$$

```python
sum_4 = Sum(x**k / factorial(k), (k, 0, oo))
result_4 = sum_4.doit()
print(f"The symbolic expression for the sum is: {sum_4}")
print(f"Symbolically valuated result: {result_4}\n")
```
