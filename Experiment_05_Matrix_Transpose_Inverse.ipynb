# Experiment 5

## Transpose of a matrix

### Objectives

1. To define a $3\times 3$ matrix $M$ with random integers between 0 and 2 as elements.
1. To find the transpose of $M$.
1. To verify that the product of $M$ and its transpose is a symmetric matrix.

### Importing required libraries

```python
import numpy as np
```

### Defining the matrix

```python
M = np.random.randint(0, 3, size=(3, 3))
print("Original Matrix M:")
print(M)
```

### Finding transpose of the matrix

```python
M_T = M.T
print("\nTranspose of M (M_T):")
print(M_T)
```

### Computing product of matrix and its transpose

```python
P = np.dot(M, M_T)
print("\nProduct P = M * M_T:")
print(P)
```

### Verification of symmetry of product matrix

```python
P_T = P.T
print("\nTranspose of P (P_T):")
print(P_T)

is_symmetric = np.array_equal(P, P_T)
print(f"\nIs the product matrix P symmetric? {is_symmetric}")

if is_symmetric:
    print("Verification successful: The product of M and its transpose is a symmetric matrix.")
else:
    print("Verification failed: The product of M and its transpose is NOT a symmetric matrix.")
```
