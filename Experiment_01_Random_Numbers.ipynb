# Experiment 1

## Uniformly distributed random numbers

### Objectives

1. To generate 100 random numbers from a uniform distribution using the `random()` function in Numpy.
1. To display these random numbers on a scatter plot.
1. To visualize the distribution of these random numbers on a histogram.
1. To compute the average and the standard deviation of these random numbers.

### Importing required libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Generating random numbers

```python
N = 200
x = np.random.random(N)
```

```python
len(x)
```

### Scatter plot of random numbers

```python
plt.figure(figsize=(10, 5))
plt.scatter(range(len(x)), x)
plt.title(f'Scatter Plot of {len(x)} Uniformly Distributed Random Numbers')
plt.xlabel('Index')
plt.ylabel('Value')
plt.grid(True)
plt.show()
```

### Histogram of random number distribution

```python
n_bins = 10

plt.figure(figsize=(10, 5))
plt.hist(x, bins=n_bins, edgecolor='black')
plt.title(f'Histogram of {len(x)} Uniformly Distributed Random Numbers')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

### Average and standard deviation of random numbers

```python
average = np.mean(x)
std_dev = np.std(x)

print(f"Average of the {len(x)} random numbers: {average:.4f}")
print(f"Standard deviation of the {len(x)} random numbers: {std_dev:.4f}")
```
